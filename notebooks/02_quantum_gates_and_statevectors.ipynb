# Quantum Gates and Statevectors

Explore phase gates, controlled-Z, SWAP, and controlled-SWAP using exact statevectors and circuit diagrams.

Run the cells from top to bottom in a fresh Python kernel. See the [repository README](../README.md) for environment setup.

**Bit ordering:** Qiskit displays bit strings with the highest-index bit on the left and bit 0 on the right. Ideal simulator results are used here; shot counts can fluctuate when several outcomes have nonzero probability.

```python
# Imports needed to run this notebook independently.
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

## Converting phase into a bit flip: H-Z-H

The identity HZH = X maps the initial |0> state to |1>. All ideal measurement shots return `1`.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
from qiskit.quantum_info import Statevector

qs = QuantumCircuit(1)

qs.h(0)
qs.z(0)
qs.h(0)

vtr = Statevector.from_instruction(qs)
print(vtr)
print(vtr.probabilities())
print(vtr.probabilities_dict())

qs.measure_all()
sm = Aer.get_backend("qasm_simulator")
tc = transpile(qs , sm)
job = sm.run(tc , shots = 1000)
rs = job.result()

cnts = rs.get_counts(qs)

display(qs.draw("mpl"))
print(cnts)
display(plot_histogram(cnts))
```

## Four T gates

T applied four times equals Z. On the prepared |1> state the result is -|1>; the global phase does not change measurement probabilities.

```python
qc = QuantumCircuit(1)

# Prepare |1>
qc.x(0)

qc.t(0)
qc.t(0)
qc.t(0)
qc.t(0)

state = Statevector.from_instruction(qc)

print(state)
print(state.probabilities_dict())

display(qc.draw("mpl"))
```

## Controlled-Z

CZ changes the sign of |11>. For this basis-state input the sign is a global phase, so the probability of `11` remains 1.

```python
qc = QuantumCircuit(2)

# Prepare |11>
qc.x(0)
qc.x(1)

qc.cz(0, 1)

state = Statevector.from_instruction(qc)

print(state)
print(state.probabilities_dict())

display(qc.draw("mpl"))
```

## SWAP

The initial displayed state `01` becomes `10` when q0 and q1 are exchanged.

```python
qc = QuantumCircuit(2)

# q0 = 1, q1 = 0
qc.x(0)

print("Before SWAP:")
print(Statevector.from_instruction(qc).probabilities_dict())

qc.swap(0, 1)

state = Statevector.from_instruction(qc)

print("After SWAP:")
print(state)
print(state.probabilities_dict())

display(qc.draw("mpl"))
```

## Controlled-SWAP (Fredkin gate)

With q0 set to 1, exchange q1 and q2. In q2q1q0 order the initial `101` becomes `011`.

```python
qc = QuantumCircuit(3)

# control q0 = 1
qc.x(0)

# q1 = 0

# q2 = 1
qc.x(2)

print("Before:")
print(Statevector.from_instruction(qc).probabilities_dict())

# q0 controls the swap of q1 and q2
qc.cswap(0, 1, 2)

state = Statevector.from_instruction(qc)

print("After:")
print(state)
print(state.probabilities_dict())

display(qc.draw("mpl"))
```
